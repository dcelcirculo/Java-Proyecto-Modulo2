# Unidad 2: PrestaYa Fintech

<p>Construir un sistema que reciba solicitudes de crédito, las evalúe por puntaje, apruebe o rechace, registre las aprobadas y muestre la cartera total con orientación a objetos moderna</p>

#### A. Identidad del solicitante

- **Uso de `BigDecimal`**: Obligatorio para todos los campos de moneda:
	<ol>
		<li> `requestedAmount`, </li>
		<li> `interestRate`, </li>
		<li> `monthlyInstallment`.</li>
	</ol>
	
	Se debe configurar con `RoundingMode.HALF_UP`.<br>

- **Record `Customer`**:
Representa al solicitante.
Debe contener:
	<ol>
		<li> `id` (String), </li>
		<li> `name` (String), </li> <li> `creditScore` (int) </li>
		<li> `city` (String). </li>
	</ol>

#### B. Jerarquía de Créditos

Modelar los diferentes productos financieros que ofrece la plataforma:

- **Clase Abstracta `Loan`**: Campos: 
	<ol>
		<li>`loanId`, </li>
		<li> `amount`, </li>
		<li> `termMonths` </li>
	</ol>
	<br>
	Método abstracto `calculateMonthlyInstallment()`.
	<br>
	<br>

- **Subclase `EntrepreneurLoan`**: Destinado a negocios. Tiene una tasa de interés preferencial fija del **1.2%** mensual.
<br>

- **Subclase `SocialHousingLoan`**: Destinado a mejoras de hogar. Incluye un campo adicional `governmentSubsidieAmount` que se resta al monto total antes de calcular la cuota.

#### C. Sistema de Evaluación

Para manejar la respuesta de una solicitud de crédito, se debe implementar una **Interfaz Sellada** llamada `ApplicationStatus`:
<br>

1. **Record `ApprovedStatus`**: Contiene:
	<ol>
		<li> `approvedAmount` </li>
		<li> `calculatedInstallment`. </li>
	</ol>

2. **Record `RejectedStatus`**: Contiene el `rejectionReason` (Ej: "Puntaje de crédito insuficiente").
<br>

Se debe usar un **Switch con Pattern Matching** para imprimir la respuesta adecuada al usuario en la consola.


#### D. Cartera de Créditos

- **Clase `PortfolioManager`**: Debe contener un **Array de objetos** `Loan` para almacenar las solicitudes procesadas con éxito (capacidad fija de 10 elementos).
<br>

- **Métodos**:
    <ol>
        <li> `registerLoan(Loan loan)`, </li>
        <li> `showPortfolio()` </li>
        <li> `calculateTotalPlacedAmount()` (suma de todos los montos de créditos aprobados). </li>
    </ol>

### Funcionalidades del Menú (Consola)

1. **Simular Nueva Solicitud (Simulate New Application)**:
    - Solicitar datos del cliente y tipo de crédito.
    - Evaluar según el puntaje de Datacrédito: si el puntaje es menor a 600, generar un `RejectedStatus`.<br>
2. **Aprobar y Registrar (Approve and Register)**:
    - Si la simulación es positiva, crear el objeto (`EntrepreneurLoan` o `SocialHousingLoan`) y guardarlo en el array de `PortfolioManager`.<br>
3. **Ver Cartera Total (View Total Portfolio)**:
    - Listar todos los créditos aprobados mostrando el detalle específico de cada uno mediante polimorfismo (`toString()`).<br>
4. **Estado de la Plataforma (Platform Status)**:
    - Uso de un **Enum** `PlatformStatus` 
		<ol>
			<li> `OPERATIONAL`, </li>
			<li> `UNDER_MAINTENANCE`, </li>
			<li> `OUT_OF_SERVICE`. </li>
		</ol>

### Requerimientos Técnicos (Obligatorios)

1. **Precisión Bancaria**: Implementar fórmulas financieras usando los métodos de la clase `BigDecimal` (`add`, `subtract`, `multiply`, `divide`).<br>

2. **Inmutabilidad**: Los datos del cliente y los resultados de la solicitud deben ser inmutables.<br>

3. **Validación de Array**: Validar que no se exceda la capacidad del array de cartera y mostrar un mensaje de error profesional si sucede.<br>

4. **Formateo de Salida**: Usar **Text Blocks** para mostrar el "Plan de Pagos" resumido al cliente.

## **Definiciones importantes**

### **Record**
Un `record` es un tipo especial de clase en Java diseñado para representar **datos inmutables** de forma simple.
Se utiliza cuando un objeto solo necesita almacenar información y no cambiar después de ser creado.
Java genera automáticamente el constructor, getters y métodos básicos, reduciendo código y evitando errores.

---

### **Sealed Interface**
Una `sealed interface` es una interfaz que restringe qué clases pueden implementarla.
Permite definir un conjunto **cerrado y controlado de tipos**, lo que mejora la seguridad y claridad del diseño.
En este proyecto, se usa para asegurar que el resultado de una solicitud solo pueda ser aprobado o rechazado.

---

### **Pattern Matching**
El `pattern matching` es una característica que permite trabajar directamente con el tipo real de un objeto sin hacer conversiones manuales.
Se usa comúnmente con `switch`, permitiendo identificar el tipo (por ejemplo, aprobado o rechazado) y acceder a sus datos de forma más limpia y segura.

---

### **Text Blocks**
Los `text blocks` son una forma de escribir texto multilínea en Java de manera más legible.
Permiten incluir textos largos (como mensajes o salidas en consola) sin necesidad de concatenar cadenas o usar múltiples comillas, mejorando la claridad del código.
